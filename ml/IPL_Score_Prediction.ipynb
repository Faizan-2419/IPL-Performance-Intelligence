```python
import pandas as pd
import numpy as np

ipl_ball_ball = pd.read_csv("ipl_ball_ball.csv", low_memory=False)
ipl_matches = pd.read_csv("ipl_matches.csv")

ipl_ball_ball.shape, ipl_matches.shape
```

```python
ipl_ball_ball.info()
ipl_matches.info()

display(ipl_ball_ball.head())
display(ipl_matches.head())
```

```python
ml_data = ipl_ball_ball[
    (ipl_ball_ball["innings"] == 1) &
    (~ipl_ball_ball["is_super_over"])
].copy()

ml_data = ml_data.sort_values(
    ["match_id", "over", "delivery_in_over"],
    kind="stable"
).reset_index(drop=True)

valid_matches = ipl_matches.loc[
    (ipl_matches["result"] == "Normal") &
    (ipl_matches["method"].isna()) &
    (ipl_matches["first_innings_runs"].notna()),
    ["match_id", "first_innings_runs", "date"]
].drop_duplicates("match_id")

ml_data = ml_data.merge(
    valid_matches,
    on="match_id",
    how="inner",
    suffixes=("", "_match")
).rename(columns={"first_innings_runs": "final_score"})

ml_data.shape
```

```python
df = ml_data.copy()

checkpoints = (
    df[df["legal_balls_bowled"].between(30, 114)]
    .sort_values(["match_id", "over", "delivery_in_over"], kind="stable")
    .drop_duplicates(["match_id", "legal_balls_bowled"], keep="last")
    .copy()
)

checkpoints = checkpoints[
    checkpoints["legal_balls_bowled"] % 6 == 0
].copy()

checkpoints["current_score"] = checkpoints["team_score"]
checkpoints["wickets_lost"] = checkpoints["team_wickets"]
checkpoints["overs_completed"] = checkpoints["legal_balls_bowled"] / 6
checkpoints["overs_remaining"] = 20 - checkpoints["overs_completed"]

checkpoints["current_run_rate"] = (
    checkpoints["current_score"] / checkpoints["overs_completed"]
)

history = df[
    ["match_id", "legal_balls_bowled", "team_score", "team_wickets"]
].drop_duplicates(["match_id", "legal_balls_bowled"], keep="last").copy()

history["legal_balls_bowled"] += 30

history = history.rename(columns={
    "team_score": "score_5_overs_ago",
    "team_wickets": "wickets_5_overs_ago"
})

checkpoints = checkpoints.merge(
    history,
    on=["match_id", "legal_balls_bowled"],
    how="left"
)

checkpoints["last_5_over_runs"] = (
    checkpoints["current_score"] -
    checkpoints["score_5_overs_ago"].fillna(0)
)

checkpoints["last_5_over_wickets"] = (
    checkpoints["wickets_lost"] -
    checkpoints["wickets_5_overs_ago"].fillna(0)
)

features = [
    "match_id", "season", "date",
    "batting_team", "bowling_team",
    "current_score", "wickets_lost",
    "overs_completed", "overs_remaining",
    "current_run_rate", "last_5_over_runs",
    "last_5_over_wickets", "final_score"
]

model_df = checkpoints[features].copy()

display(model_df.head())
```

```python
df = ml_data.sort_values(
    ["match_id", "over", "delivery_in_over"],
    kind="stable"
).copy()

states = df.drop_duplicates(
    ["match_id", "legal_balls_bowled"],
    keep="last"
).copy()

checkpoints = states[
    states["legal_balls_bowled"].between(30, 114)
    & states["legal_balls_bowled"].mod(6).eq(0)
].copy()

previous = states[
    ["match_id", "legal_balls_bowled", "team_score", "team_wickets"]
].copy()

previous["legal_balls_bowled"] += 30

previous = previous.rename(columns={
    "team_score": "previous_score",
    "team_wickets": "previous_wickets"
})

checkpoints = checkpoints.merge(
    previous,
    on=["match_id", "legal_balls_bowled"],
    how="left",
    validate="one_to_one"
)

checkpoints["current_score"] = checkpoints["team_score"]
checkpoints["wickets_lost"] = checkpoints["team_wickets"]
checkpoints["overs_completed"] = checkpoints["legal_balls_bowled"] / 6
checkpoints["overs_remaining"] = 20 - checkpoints["overs_completed"]

checkpoints["current_run_rate"] = (
    checkpoints["current_score"] / checkpoints["overs_completed"]
)

checkpoints["last_5_over_runs"] = (
    checkpoints["current_score"] - checkpoints["previous_score"].fillna(0)
)

checkpoints["last_5_over_wickets"] = (
    checkpoints["wickets_lost"] - checkpoints["previous_wickets"].fillna(0)
)

model_df = checkpoints[features].copy()

display(model_df.head())
```

```python
eligible_matches = ipl_matches.loc[
    (ipl_matches["result"] == "Normal") &
    (ipl_matches["method"].isna()) &
    (
        (ipl_matches["first_innings_overs"] >= 19.5) |
        (ipl_matches["first_innings_wickets"] == 10)
    ),
    "match_id"
].unique()

model_df = model_df[
    model_df["match_id"].isin(eligible_matches)
].copy()

model_df["date"] = pd.to_datetime(model_df["date"], errors="coerce")

match_order = (
    model_df[["match_id", "date"]]
    .drop_duplicates("match_id")
    .sort_values(["date", "match_id"])
)

split_index = int(len(match_order) * 0.8)

train_ids = match_order.iloc[:split_index]["match_id"]
test_ids = match_order.iloc[split_index:]["match_id"]

train_df = model_df[model_df["match_id"].isin(train_ids)].copy()
test_df = model_df[model_df["match_id"].isin(test_ids)].copy()

feature_columns = [
    "batting_team",
    "bowling_team",
    "current_score",
    "wickets_lost",
    "overs_completed",
    "overs_remaining",
    "current_run_rate",
    "last_5_over_runs",
    "last_5_over_wickets"
]

X_train = train_df[feature_columns]
X_test = test_df[feature_columns]

y_train = train_df["final_score"]
y_test = test_df["final_score"]

train_df.shape, test_df.shape
```

```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

categorical_features = ["batting_team", "bowling_team"]

numerical_features = [
    col for col in feature_columns
    if col not in categorical_features
]

preprocessor = ColumnTransformer([
    ("teams", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ("numerical", "passthrough", numerical_features)
])

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        max_depth=15,
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    ))
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_predictions))
rf_r2 = r2_score(y_test, rf_predictions)

rf_metrics = pd.DataFrame({
    "MAE": [rf_mae],
    "RMSE": [rf_rmse],
    "R2": [rf_r2]
}).round(4)

display(rf_metrics)

results_df = test_df[
    [
        "match_id", "season", "batting_team",
        "bowling_team", "overs_completed",
        "current_score", "final_score"
    ]
].copy()

results_df["predicted_score"] = rf_predictions
results_df["absolute_error"] = (
    results_df["final_score"] - results_df["predicted_score"]
).abs()

display(results_df.head(10))
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Existing test predictions
evaluation_df = results_df.copy()

# Simple baseline:
# Project the current run rate over 20 overs
evaluation_df["baseline_score"] = (
    evaluation_df["current_score"] /
    evaluation_df["overs_completed"] * 20
)

# Prediction must not be below runs already scored
evaluation_df["baseline_score"] = np.maximum(
    evaluation_df["baseline_score"],
    evaluation_df["current_score"]
)

evaluation_df["rf_prediction"] = np.maximum(
    evaluation_df["predicted_score"],
    evaluation_df["current_score"]
)

# Compare across innings checkpoints
records = []

for over in [5, 8, 10, 12, 15, 18, 19]:
    subset = evaluation_df[
        evaluation_df["overs_completed"] == over
    ]

    if subset.empty:
        continue

    records.append({
        "Over": over,
        "Matches": len(subset),
        "RF_MAE": mean_absolute_error(
            subset["final_score"],
            subset["rf_prediction"]
        ),
        "Baseline_MAE": mean_absolute_error(
            subset["final_score"],
            subset["baseline_score"]
        ),
        "RF_RMSE": np.sqrt(mean_squared_error(
            subset["final_score"],
            subset["rf_prediction"]
        ))
    })

phase_results = pd.DataFrame(records).round(2)

display(phase_results)

print("\nOVERALL COMPARISON")
print(
    "Random Forest MAE:",
    round(mean_absolute_error(
        evaluation_df["final_score"],
        evaluation_df["rf_prediction"]
    ), 2)
)
print(
    "Run-rate baseline MAE:",
    round(mean_absolute_error(
        evaluation_df["final_score"],
        evaluation_df["baseline_score"]
    ), 2)
)
```

```python

from sklearn.ensemble import HistGradientBoostingRegressor

gb_preprocessor = ColumnTransformer([
    ("teams", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_features),
    ("numerical", "passthrough", numerical_features)
])

gb_pipeline = Pipeline([
    ("preprocessor", gb_preprocessor),
    ("model", HistGradientBoostingRegressor(
        max_iter=300,
        learning_rate=0.05,
        max_leaf_nodes=20,
        l2_regularization=5.0,
        random_state=42
    ))
])

gb_pipeline.fit(X_train, y_train)

gb_predictions = np.maximum(
    gb_pipeline.predict(X_test),
    X_test["current_score"].to_numpy()
)

rf_predictions = np.maximum(
    rf_predictions,
    X_test["current_score"].to_numpy()
)

baseline_predictions = (
    X_test["current_score"] / X_test["overs_completed"] * 20
)

comparison = pd.DataFrame([
    {
        "Model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2": r2_score(y_test, pred)
    }
    for name, pred in [
        ("Random Forest", rf_predictions),
        ("Gradient Boosting", gb_predictions),
        ("Run-Rate Baseline", baseline_predictions)
    ]
]).round(3)

display(comparison.sort_values("MAE"))
```

```python

gb_results_df = test_df[
    [
        "match_id", "season", "date",
        "batting_team", "bowling_team",
        "overs_completed", "current_score",
        "wickets_lost", "final_score"
    ]
].copy()

gb_results_df["predicted_score"] = gb_predictions

gb_results_df["absolute_error"] = (
    gb_results_df["final_score"] - gb_results_df["predicted_score"]
).abs()

gb_over_results = (
    gb_results_df
    .groupby("overs_completed")
    .agg(
        Matches=("match_id", "nunique"),
        MAE=("absolute_error", "mean"),
        RMSE=("absolute_error", lambda x: np.sqrt(np.mean(x ** 2)))
    )
    .reset_index()
    .round(2)
)

display(gb_over_results)

gb_metrics = pd.DataFrame({
    "MAE": [mean_absolute_error(y_test, gb_predictions)],
    "RMSE": [np.sqrt(mean_squared_error(y_test, gb_predictions))],
    "R2": [r2_score(y_test, gb_predictions)]
}).round(4)

display(gb_metrics)
```

```python

# STEP 10: Audit match eligibility and innings lengths

audit = ipl_matches[
    ipl_matches["match_id"].isin(model_df["match_id"])
].copy()

# Cricket overs notation: 19.5 = 19 overs + 5 balls
audit["overs_notation"] = audit["first_innings_overs"]

audit["legal_balls_from_overs"] = (
    np.floor(audit["overs_notation"]).astype(int) * 6
    + np.rint(
        (audit["overs_notation"] % 1) * 10
    ).astype(int)
)

# Identify unusually short innings
audit["possibly_shortened"] = (
    (audit["legal_balls_from_overs"] < 120)
    & (audit["first_innings_wickets"] < 10)
)

print("Total selected matches:", len(audit))
print("Possibly shortened innings:",
      audit["possibly_shortened"].sum())

print("\nFirst-innings legal-ball distribution:")
print(
    audit["legal_balls_from_overs"]
    .value_counts()
    .sort_index()
    .tail(15)
)

print("\nPotentially problematic matches:")
display(
    audit.loc[
        audit["possibly_shortened"]
        ,
        [
            "match_id",
            "season",
            "first_innings_runs",
            "first_innings_wickets",
            "first_innings_overs",
            "target_overs",
            "method",
            "result"
        ]
    ].head(20)
)

print("\nInnings exceeding 120 legal balls:",
      (audit["legal_balls_from_overs"] > 120).sum())
```

```python

invalid_match_ids = audit.loc[
    audit["possibly_shortened"] |
    (audit["legal_balls_from_overs"] > 120),
    "match_id"
].unique()

final_model_df = model_df[
    ~model_df["match_id"].isin(invalid_match_ids)
].copy()

match_order = (
    final_model_df[["match_id", "date"]]
    .drop_duplicates("match_id")
    .sort_values(["date", "match_id"])
)

split_index = int(len(match_order) * 0.8)

train_ids = match_order.iloc[:split_index]["match_id"]
test_ids = match_order.iloc[split_index:]["match_id"]

train_df = final_model_df[
    final_model_df["match_id"].isin(train_ids)
].copy()

test_df = final_model_df[
    final_model_df["match_id"].isin(test_ids)
].copy()

X_train = train_df[feature_columns]
X_test = test_df[feature_columns]

y_train = train_df["final_score"]
y_test = test_df["final_score"]

gb_pipeline.fit(X_train, y_train)

gb_predictions = np.maximum(
    gb_pipeline.predict(X_test),
    X_test["current_score"].to_numpy()
)

final_mae = mean_absolute_error(y_test, gb_predictions)
final_rmse = np.sqrt(mean_squared_error(y_test, gb_predictions))
final_r2 = r2_score(y_test, gb_predictions)

final_metrics = pd.DataFrame({
    "MAE": [final_mae],
    "RMSE": [final_rmse],
    "R2": [final_r2]
}).round(4)

display(final_metrics)
```

```python

import joblib
from pathlib import Path

output_dir = Path("IPL_Score_Prediction")
models_dir = output_dir / "models"
powerbi_dir = output_dir / "powerbi_data"

models_dir.mkdir(parents=True, exist_ok=True)
powerbi_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(gb_pipeline, models_dir / "ipl_score_model.pkl")
joblib.dump(feature_columns, models_dir / "feature_columns.pkl")

prediction_df = test_df[
    [
        "match_id", "season", "date",
        "batting_team", "bowling_team",
        "overs_completed", "current_score",
        "wickets_lost", "final_score"
    ]
].copy()

prediction_df["predicted_score"] = gb_predictions

prediction_df["absolute_error"] = (
    prediction_df["final_score"] - prediction_df["predicted_score"]
).abs()

prediction_df["prediction_error"] = (
    prediction_df["predicted_score"] - prediction_df["final_score"]
)

prediction_df["match_phase"] = pd.cut(
    prediction_df["overs_completed"],
    bins=[0, 6, 15, 20],
    labels=["Powerplay", "Middle", "Death"]
)

metrics_df = pd.DataFrame([{
    "model_name": "HistGradientBoostingRegressor",
    "model_version": "1.0",
    "MAE": final_mae,
    "RMSE": final_rmse,
    "R2": final_r2,
    "training_matches": train_df["match_id"].nunique(),
    "testing_matches": test_df["match_id"].nunique()
}])

over_metrics_df = (
    prediction_df
    .groupby("overs_completed", observed=True)
    .agg(
        matches=("match_id", "nunique"),
        MAE=("absolute_error", "mean"),
        RMSE=("prediction_error", lambda x: np.sqrt(np.mean(x ** 2)))
    )
    .reset_index()
)

prediction_df.to_csv(powerbi_dir / "ipl_predictions.csv", index=False)
metrics_df.to_csv(powerbi_dir / "model_metrics.csv", index=False)
over_metrics_df.to_csv(powerbi_dir / "over_performance.csv", index=False)

display(metrics_df)
```
